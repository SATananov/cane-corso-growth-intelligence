# 14 — Baseline Image Classifier Training Plan

This notebook documents the planned first baseline image classifier for the Cane Corso Growth Intelligence visual-similarity extension.

It does not train a model yet. It explains the model objective, data splits, metrics, and interpretation boundaries.


## Objective

The future baseline model should estimate visual similarity among the selected local Stanford Dogs baseline classes.

The model output should be interpreted as a probability distribution over the trained classes, not as a proof of breed, pedigree, registry identity, genetics, or health status.


## Planned training pipeline

1. Load the local baseline subset manifest.
2. Read images from the local `train`, `validation`, and `test` split folders.
3. Resize images to a fixed input size, for example 224 × 224.
4. Use a pretrained image backbone.
5. Train a small classification head or use extracted features with a simple classifier.
6. Evaluate with accuracy, macro F1, per-class precision/recall, and a confusion matrix.
