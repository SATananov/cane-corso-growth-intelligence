### Visualization 1 — SVD text-note space

This 2D view uses the first two SVD components from TF-IDF growth notes. It is useful for checking whether healthy, monitor and risk notes separate in the latent text space.


```python
import matplotlib.pyplot as plt

viz = pd.read_csv(report_dir / "dimensionality_reduction_problem5_visualization_coordinates.csv")
svd_viz = viz[viz["visualization"] == "SVD_2D_text_notes"]

fig, ax = plt.subplots(figsize=(7, 5))
for label, group in svd_viz.groupby("label"):
    ax.scatter(group["x"], group["y"], label=label)
ax.set_title("Problem 5 — SVD 2D text-note space")
ax.set_xlabel("SVD component 1")
ax.set_ylabel("SVD component 2")
ax.legend()
plt.show()
```

### Visualization 2 — PCA numeric reference space

This reference view demonstrates the lecture concept of a 2D PCA projection after scaling. It is included as a second visualization while keeping the repository free of external datasets.


```python
pca_viz = viz[viz["visualization"] == "PCA_2D_numeric_reference"]

fig, ax = plt.subplots(figsize=(7, 5))
for label, group in pca_viz.groupby("label"):
    ax.scatter(group["x"], group["y"], label=label)
ax.set_title("Problem 5 — PCA 2D numeric reference space")
ax.set_xlabel("PCA component 1")
ax.set_ylabel("PCA component 2")
ax.legend()
plt.show()
```
