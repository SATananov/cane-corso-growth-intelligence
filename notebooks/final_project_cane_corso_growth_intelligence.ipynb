## 1. Project Motivation

Large-breed puppies can grow quickly and unevenly. A single measurement is not enough to understand the full development story. The project therefore treats growth monitoring as a sequence of structured records.

The practical motivation is to help an owner ask better data questions:

- What is the expected weight for the current age and body measurements?
- Is the current record close to the expected pattern?
- Does the record look like normal growth or a signal that deserves closer attention?
- Which growth-pattern group does the dog resemble?
- How does the growth trajectory change over time?

The goal is not to replace professional judgment. The goal is to transform simple records into interpretable machine-learning signals.


## 2. Problem Definition

The project uses the same real-world domain and applies several machine-learning perspectives to it.

| ML task | Project question | Output |
|---|---|---|
| Regression | What is the expected weight/growth value? | numerical prediction |
| Classification | Does the record indicate normal growth or a possible attention signal? | probability/class |
| Clustering | Which natural growth-pattern group does the record resemble? | segment/cluster |
| Feature Engineering | Which derived variables explain growth better? | ratios, lags, velocity, rolling statistics |
| Time Series | How does growth evolve through ordered measurements? | trajectory features and trend signals |
| Dimensionality Reduction | Can high-dimensional growth records and text-note representations be projected and interpreted? | lower-dimensional components / visual coordinates |
| Computer Vision Extension | Can public images support an educational visual-similarity prototype? | exploratory plan/prototype only |

Machine Learning Tools remains reserved as a future course-aligned addition after the lecture and exercise are studied in class.

## 4. Mathematical Framing

Each growth record can be represented as a feature vector:

$$
x = [age\_months, weight\_kg, height\_cm, sex\_encoded, body\_ratio, growth\_velocity, deviation\_from\_expected]
$$

For regression, the model learns a numerical function:

$$
\hat{y} = f(x)
$$

The residual explains prediction error:

$$
residual = y - \hat{y}
$$

For classification, the model estimates a probability:

$$
P(needs\_attention \mid x)
$$

For clustering, the model searches for natural structure without labels:

$$
cluster\_id = g(x)
$$

For time-series analysis, each dog can be represented as an ordered trajectory:

$$
trajectory = [(age_1, weight_1), (age_2, weight_2), ..., (age_n, weight_n)]
$$

## 5. Linear Regression for Growth Prediction

The regression part models expected bodyweight or growth values from growth-related features. The main learning idea is:

$$
\hat{y}_{weight} = f(age, height, sex, activity, engineered\ features)
$$

Evaluation focuses on numerical error: **MAE**, **RMSE** and **R²**. The regression work is supported by the main regression notebook and the course-alignment report files.


````python
def read_csv_report(relative_path: str) -> pd.DataFrame:
    path = ROOT / relative_path
    if not path.exists():
        print(f"Missing report: {relative_path}")
        return pd.DataFrame()
    return pd.read_csv(path)

regression_metrics = read_csv_report("reports/course_exercises/linear_regression_exercise_alignment_metrics.csv")
if not regression_metrics.empty:
    display(regression_metrics)
````

## 6. Classification for Growth Status

The classification part transforms the project into a probability-based monitoring signal:

$$

$$

This is not a medical diagnosis. It is a learning-oriented signal that demonstrates binary classification, model evaluation, threshold interpretation, and responsible communication.


````python
classification_metrics = read_csv_report("reports/course_exercises/classification_exercise_alignment_metrics.csv")
classification_ablation = read_csv_report("reports/course_exercises/classification_exercise_ablation_study.csv")
classification_confusion = read_csv_report("reports/course_exercises/classification_exercise_confusion_matrix.csv")

if not classification_metrics.empty:
    display(Markdown("### Classification model comparison"))
    display(classification_metrics)
if not classification_ablation.empty:
    display(Markdown("### Feature ablation study"))
    display(classification_ablation)
if not classification_confusion.empty:
    display(Markdown("### Confusion matrix"))
    display(classification_confusion)
````

## 10. Results Summary

At the current stage, the strongest completed course-aligned work is the tabular ML pipeline:

1. **Regression** predicts growth-related numerical values and explains errors through residuals.
2. **Classification** converts growth records into probability-based monitoring signals.
3. **Clustering** groups records into interpretable growth-pattern segments.
4. **Feature engineering and time series** improve the representation with lag, velocity, ratio and trajectory features.
5. **Dimensionality reduction and manifold learning** reduce high-dimensional representations, compare projections and strengthen the important Problem 5 component-analysis workflow.
6. **Computer vision** is documented as an optional extension, not as the core project claim.


````python
summary_rows = []

if not regression_metrics.empty:
    best_reg = regression_metrics.sort_values("rmse").iloc[0]
    summary_rows.append({"area": "Regression", "best_or_selected_result": best_reg.get("model", "best regression model"), "key_metric": f"RMSE={best_reg.get('rmse', 'n/a')}", "interpretation": "Lower prediction error means a better expected-growth estimate."})

if not classification_metrics.empty:
    best_cls = classification_metrics.sort_values("test_f1", ascending=False).iloc[0]
    summary_rows.append({"area": "Classification", "best_or_selected_result": best_cls.get("model", "best classification model"), "key_metric": f"F1={best_cls.get('test_f1', 'n/a')}", "interpretation": "F1 balances precision and recall for the attention-signal class."})

if not clustering_metrics.empty:
    selected = clustering_metrics[clustering_metrics.get("selected_for_project", False) == True]
    row = selected.iloc[0] if not selected.empty else clustering_metrics.sort_values("silhouette", ascending=False).iloc[0]
    summary_rows.append({"area": "Clustering", "best_or_selected_result": f"{row.get('algorithm', 'cluster model')} with {row.get('cluster_count', 'n/a')} clusters", "key_metric": f"silhouette={row.get('silhouette', 'n/a')}", "interpretation": "Silhouette helps compare how separated the exploratory growth segments are."})

if not time_ablation.empty:
    selected_time = time_ablation[time_ablation.get("experiment", "") == "selected_final_model"]
    row = selected_time.iloc[0] if not selected_time.empty else time_ablation.sort_values("rmse").iloc[0]
    summary_rows.append({"area": "Feature Engineering / Time Series", "best_or_selected_result": row.get("experiment", "selected time-series model"), "key_metric": f"RMSE={row.get('rmse', 'n/a')}", "interpretation": "Engineered time and velocity features improve trajectory-aware prediction."})

summary = pd.DataFrame(summary_rows)
display(summary)
````
