# Public Image Dataset Feasibility

This notebook documents the feasibility check before any visual similarity training is attempted.

The goal is to decide whether public dog image datasets can support a future molossoid visual similarity classifier. The result must be interpreted as visual similarity, not breed proof.

This notebook is part of an optional exploratory visual-similarity extension. It is not presented as a core course requirement and it does not prove breed identity, pedigree, registry status or veterinary status.


## Mathematical / ML Framing

A future image classifier would use an image tensor as input:

```text
X_image ∈ R^(H × W × C)
```

A pretrained feature extractor would map the image to a feature vector:

```text
h = phi(X_image)
```

The classifier head would produce logits and probabilities:

```text
z = W h + b
p_i = exp(z_i) / sum_j exp(z_j)
```

The probabilities should be read as similarity scores among trained classes, not as official breed evidence.


````python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
feasibility_path = PROJECT_ROOT / 'data' / 'image_dataset_feasibility_matrix.csv'
classes_path = PROJECT_ROOT / 'data' / 'molossoid_visual_target_classes.csv'

feasibility = pd.read_csv(feasibility_path)
target_classes = pd.read_csv(classes_path)

feasibility[['dataset_name', 'approximate_classes', 'approximate_images', 'step12_decision']]
````

````python
target_classes[['class_slug', 'display_name', 'role', 'dataset_availability_status']]
````

## Decision

The project should not train a visual similarity model until the selected dataset's class list and usage terms are checked.

The next safe stage is to choose one candidate dataset, inspect its metadata locally, and confirm which target classes are actually available.


````python
required_targets = {'cane_corso', 'dogo_argentino', 'presa_canario', 'other_unknown'}
available_targets = set(target_classes['class_slug'])
missing_targets = required_targets - available_targets

assert len(feasibility) >= 4, 'Expected at least four dataset candidates'
assert not missing_targets, f'Missing target classes in planning CSV: {missing_targets}'
assert set(feasibility['repository_policy']) == {'do_not_commit_downloaded_images'}

print('Notebook feasibility check PASS')
print(f'Dataset candidates: {len(feasibility)}')
print(f'Target classes in plan: {len(target_classes)}')
````

## Responsible Limitations

This feasibility stage does not download images, does not train a model, and does not claim that the target classes are already available.

If a public dataset does not contain Cane Corso or another target class, the first baseline should use only available classes and clearly document the limitation.
