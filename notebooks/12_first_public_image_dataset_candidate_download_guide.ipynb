# First Public Image Dataset Candidate / Download Guide

This notebook documents the first public image dataset candidate for the future visual similarity module. It does not download images and does not train an image model.

This notebook is part of an optional exploratory visual-similarity extension. It is not presented as a core course requirement and it does not prove breed identity, pedigree, registry status or veterinary status.


## Responsible framing

The future model should be described as a **visual similarity classifier**, not breed proof. A photo cannot prove pedigree, registry status, genetic origin, or veterinary status.


## Mathematical view

An image can be represented as a tensor:

```text
x_image ∈ R^(H × W × 3)
```

A future classifier will produce a probability vector over confirmed classes:

```text
p = [P(class_1 | x), P(class_2 | x), ..., P(class_k | x)]
```

this stage only prepares the dataset candidate; it does not estimate `p`.
