# Baseline Image Subset Preparation

This notebook documents the local preparation stage for the first Stanford Dogs visual-similarity baseline subset.

The goal is not to train the image model yet. The goal is to prepare a small, reproducible train/validation/test subset from the selected Stanford Dogs classes.

This notebook is part of an optional exploratory visual-similarity extension. It is not presented as a core course requirement and it does not prove breed identity, pedigree, registry status or veterinary status.


````python
from pathlib import Path
import sys


def find_project_root(start=None):
    start = Path.cwd() if start is None else Path(start)
    for candidate in [start, *start.parents]:
        if (candidate / "src").exists() and (candidate / "data").exists():
            return candidate
    raise RuntimeError("Project root not found. Open this notebook from the project folder or adjust the working directory.")


PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(f"Notebook ready. Project root: {PROJECT_ROOT}")
````
