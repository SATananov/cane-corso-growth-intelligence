# Step 14 — Public Image Dataset Class Availability Audit

This notebook documents the class-availability check for the future Molossoid Visual Similarity module.


## Boundary

This is not a training notebook. It checks whether public image datasets may contain useful classes for a future visual similarity classifier. The output must never be interpreted as breed proof, pedigree proof, or veterinary evidence.


```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
candidates = pd.read_csv(project_root / 'data' / 'public_image_dataset_class_candidates.csv')
rules = pd.read_csv(project_root / 'data' / 'public_image_dataset_target_audit_rules.csv')

candidates.head()
```

```python
print('Candidate rows by dataset:')
display(candidates.groupby('dataset_name').size().rename('rows').reset_index())

print('Candidate rows by target class:')
display(candidates.groupby('target_class').size().rename('rows').reset_index())

print('Step 15 usability flags:')
display(candidates.groupby('usable_for_step15_baseline').size().rename('rows').reset_index())
```

```python
import subprocess, sys

script = project_root / 'src' / 'audit_public_image_dataset_classes.py'
result = subprocess.run([sys.executable, str(script)], cwd=project_root, text=True, capture_output=True)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise SystemExit(result.returncode)
```

## Conclusion

The next baseline image classifier should use only confirmed available public classes. Cane Corso-specific visual classification should remain future work unless a clean public Cane Corso class or a consent-based USG image dataset becomes available.
