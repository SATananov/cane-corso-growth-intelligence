## Run the compact alignment workflow

The script below compares three regression approaches:

1. age-only linear regression baseline;
2. enriched linear regression with numeric and categorical preprocessing;
3. tuned Ridge regression with a small `alpha` grid.


```python
from pathlib import Path
import subprocess
import sys


def project_root_from_notebook():
    current = Path.cwd().resolve()
    for candidate in [current, *current.parents]:
        if (candidate / "src").exists() and (candidate / "data").exists():
            return candidate
    raise RuntimeError("Project root not found. Open the notebook from the project folder.")


PROJECT_ROOT = project_root_from_notebook()


def run_project_script(relative_script):
    script_path = PROJECT_ROOT / relative_script
    result = subprocess.run(
        [sys.executable, str(script_path)],
        cwd=PROJECT_ROOT,
        text=True,
        capture_output=True,
    )
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr)
    result.check_returncode()
    return result.returncode


run_project_script(Path("src/run_linear_regression_exercise_alignment.py"))

```
