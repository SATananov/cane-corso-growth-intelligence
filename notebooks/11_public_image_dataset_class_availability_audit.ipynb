```python
from pathlib import Path
import os
import sys

def find_project_root(start=None):
    current = Path(start or Path.cwd()).resolve()
    for candidate in [current, *current.parents]:
        if (
            (candidate / "README.md").exists()
            and (candidate / "src").is_dir()
            and (candidate / "data").is_dir()
            and (candidate / "notebooks").is_dir()
        ):
            return candidate
    raise RuntimeError("Project root was not found. Open this notebook from inside the project folder.")

PROJECT_ROOT = find_project_root()
ROOT = PROJECT_ROOT
root = PROJECT_ROOT
project_root = PROJECT_ROOT

os.chdir(PROJECT_ROOT)

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
print("Working directory:", Path.cwd())
```

# Public Image Dataset Class Availability Audit

This notebook documents the class-availability check for the future Molossoid Visual Similarity module.

This notebook is part of an optional exploratory visual-similarity extension. It is not presented as a core course requirement and it does not prove breed identity, pedigree, registry status or veterinary status.


## Boundary

This is not a training notebook. It checks whether public image datasets may contain useful classes for a future visual similarity classifier. The output must never be interpreted as breed proof, pedigree proof, or veterinary evidence.


```python
from pathlib import Path
import pandas as pd
project_root = next(candidate for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents] if (candidate / "src").is_dir() and (candidate / "data").is_dir())
candidates = pd.read_csv(project_root / 'data' / 'public_image_dataset_class_candidates.csv')
rules = pd.read_csv(project_root / 'data' / 'public_image_dataset_target_audit_rules.csv')

candidates.head()
```
