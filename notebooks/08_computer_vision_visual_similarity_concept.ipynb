## Problem Formulation

Input:

```text
X_image = uploaded dog image
```

Target classes, if legally usable data exists:

```text
cane_corso, dogo_argentino, presa_canario, great_dane, neapolitan_mastiff, bullmastiff, boxer, mastiff, other_unknown
```

Output:

```text
P(class_i | X_image)
```

The interpretation should be: the image is visually closest to the highest-probability class among the trained classes.


## Mathematical Model Idea

A future transfer-learning model can be represented as:

```text
h = phi(X_image)
z = W h + b
p_i = exp(z_i) / sum(exp(z_j))
```

Where:

- `phi` is a pretrained visual feature extractor;
- `h` is the image embedding;
- `z` are logits;
- `p_i` are softmax probabilities.


````python
import math

def softmax(logits):
    max_logit = max(logits.values())
    exp_values = {label: math.exp(value - max_logit) for label, value in logits.items()}
    total = sum(exp_values.values())
    return {label: value / total for label, value in exp_values.items()}

example_logits = {
    'cane_corso': 2.40,
    'dogo_argentino': 1.15,
    'presa_canario': 0.95,
    'great_dane': 0.40,
    'other_unknown': -0.10,
}

probabilities = softmax(example_logits)
for label, probability in sorted(probabilities.items(), key=lambda item: item[1], reverse=True):
    print(f'{label}: {probability:.2%}')
````

## Combined Growth + Visual Report Concept

A future combined report could include both signals:

```text
Growth signal:
- growth velocity z-score
- trajectory trend
- educational monitoring status

Visual similarity signal:
- top-k breed-similarity probabilities
- confidence / uncertainty warning
- class coverage explanation
```

The system should never use the image prediction as official breed proof.


````python
growth_signal = {
    'latest_velocity_z_score': -0.87,
    'monitoring_signal': 'stable_in_learning_context',
}

top_visual_label = max(probabilities, key=probabilities.get)
combined_report = {
    'growth_signal': growth_signal,
    'top_visual_similarity': top_visual_label,
    'top_visual_probability': round(probabilities[top_visual_label], 4),
    'responsible_boundary': 'educational visual similarity only; not breed proof or veterinary diagnosis',
}

combined_report
````
