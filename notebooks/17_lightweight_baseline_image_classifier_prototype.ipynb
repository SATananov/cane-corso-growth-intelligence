# Lightweight Baseline Image Classifier Prototype

This notebook documents the first lightweight image-classification prototype for the visual-similarity extension.

This notebook is part of an optional exploratory visual-similarity extension. It is not presented as a core course requirement and it does not prove breed identity, pedigree, registry status or veterinary status.


```python
from pathlib import Path
import sys


def find_project_root(start=None):
    start = Path.cwd() if start is None else Path(start)
    for candidate in [start, *start.parents]:
        if (candidate / "src").exists() and (candidate / "data").exists():
            return candidate
    raise RuntimeError("Project root not found. Open this notebook from the project folder or adjust the working directory.")


PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(f"Notebook ready. Project root: {PROJECT_ROOT}")
```

## Boundary

The output is visual similarity among trained classes only. It is not breed proof, pedigree proof, registry proof, genetic proof, or veterinary advice.


## Local workflow

Run from the project root after the this stage baseline subset has been prepared locally.


```python
import subprocess

subset_root = PROJECT_ROOT / "data" / "images" / "local_dataset" / "splits" / "stanford_dogs_first_baseline"
image_files = []
if subset_root.exists():
    image_files = list(subset_root.rglob("*.jpg")) + list(subset_root.rglob("*.jpeg")) + list(subset_root.rglob("*.png"))

if image_files:
    subprocess.run([sys.executable, str(PROJECT_ROOT / "src" / "train_lightweight_image_classifier.py")], check=True)
    subprocess.run([sys.executable, str(PROJECT_ROOT / "src" / "validate_lightweight_image_classifier_outputs.py")], check=True)
else:
    print("Local image subset not found. Prepare the local subset before training this exploratory visual-similarity baseline.")
    print(f"Expected folder: {subset_root}")
```
