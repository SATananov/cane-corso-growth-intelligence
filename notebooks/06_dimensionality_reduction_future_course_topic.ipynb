## PCA explained variance

PCA creates new orthogonal components. The first components capture the largest amount of variance. In a real growth dataset, this can help compress correlated measurements such as weight, height, chest size, and growth deltas.


```python
pca_variance = pd.read_csv(report_dir / "dimensionality_reduction_pca_explained_variance.csv")
pca_variance
```

## PCA components

The component table shows how each original numeric feature contributes to every principal component. This is useful for interpreting whether a reduced dimension mostly represents size, proportions, or growth pattern.


```python
pca_components = pd.read_csv(report_dir / "dimensionality_reduction_pca_components.csv")
pca_components
```

## Supervised and manifold representations

LinDA uses labels and is useful for classification tasks. Isomap and t-SNE are used to inspect geometry and possible clusters. t-SNE is treated as visualization-only, not as a stable production preprocessing step.


```python
embedding_preview = pd.read_csv(report_dir / "dimensionality_reduction_embedding_preview.csv")
embedding_preview.head()
```

## Kernel PCA

Kernel PCA demonstrates a non-linear projection. It is especially relevant when the data shape cannot be well represented by a linear projection.


```python
kernel_metrics = pd.read_csv(report_dir / "dimensionality_reduction_kernel_pca_metrics.csv")
kernel_metrics
```

## Text latent spaces with TF-IDF + TruncatedSVD

The course exercise focuses strongly on text representations. Here the same idea is adapted to project-style notes: TF-IDF creates sparse text features, and TruncatedSVD compresses them into latent semantic components.


```python
text_metrics = pd.read_csv(report_dir / "dimensionality_reduction_text_representation_metrics.csv")
text_components = pd.read_csv(report_dir / "dimensionality_reduction_text_svd_components.csv")
text_metrics, text_components
```

## Conclusion

The module shows the trade-off between interpretability, compression, visualization, and model performance. For the final project, PCA/LinDA are the most directly useful for structured growth-status models, while TF-IDF + TruncatedSVD is useful for future text-note analysis. Isomap and t-SNE are best positioned as diagnostic visualization tools.
