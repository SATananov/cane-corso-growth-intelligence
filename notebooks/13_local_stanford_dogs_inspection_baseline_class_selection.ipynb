# Local Stanford Dogs Inspection / Baseline Class Selection

This notebook explains how the project chooses first-baseline visual classes from confirmed local Stanford Dogs labels. It does not train an image model.

This notebook is part of an optional exploratory visual-similarity extension. It is not presented as a core course requirement and it does not prove breed identity, pedigree, registry status or veterinary status.


## Main rule

A desired breed list is not the same as an available training-label list. The project must inspect local dataset labels before choosing classes.


## Classification framing

A future classifier can only produce probabilities over classes included in training:

```text
p = [P(class_1 | x_image), ..., P(class_k | x_image)]
```

Therefore, if `Cane Corso` is not in the training labels, the model cannot honestly output a Cane Corso probability. It can only compare among confirmed classes.
